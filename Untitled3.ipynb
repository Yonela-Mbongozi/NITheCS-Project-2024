```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Set the file path
file_path = "/content/states_reg_10k.csv"
```

```python
# Load the dataset
quantum_data = pd.read_csv(file_path)

# Display the first few rows of the dataset
print(quantum_data.head())

# Check the structure of the dataset
print(quantum_data.info())

# Summary statistics of the dataset
print(quantum_data.describe())
```

```python
# Check for missing values
print(f"Number of missing values: {quantum_data.isna().sum().sum()}")
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Sample data for demonstration purposes
# quantum_data = your_data_frame_here

np.random.seed(123)  # For reproducibility

# Split the data (80% train, 20% test)
train_data, test_data = train_test_split(quantum_data, test_size=0.2, random_state=123)

# Separate features and target
x_train = train_data.drop(columns=['conc']).values
y_train = train_data['conc'].values

x_test = test_data.drop(columns=['conc']).values
y_test = test_data['conc'].values

# Create individual dataframes for train and test
train_df = pd.DataFrame({'conc': y_train, 'Dataset': 'Train'})
test_df = pd.DataFrame({'conc': y_test, 'Dataset': 'Test'})

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

# Histogram with density curve for the full dataset
sns.histplot(quantum_data['conc'], kde=True, bins=30, color='red', stat="density", ax=axes[0])
axes[0].set_title("Distribution of Concurrence")
axes[0].set_xlabel("Concurrence")
axes[0].set_ylabel("Density")

# Histogram with density curve for the training dataset
sns.histplot(train_df, x="conc", kde=True, bins=30, stat="density", color="blue", ax=axes[1])
axes[1].set_title("Training Data Distribution")
axes[1].set_xlabel("Concurrence")

# Histogram with density curve for the testing dataset
sns.histplot(test_df, x="conc", kde=True, bins=30, stat="density", color="orange", ax=axes[2])
axes[2].set_title("Testing Data Distribution")
axes[2].set_xlabel("Concurrence")

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
correlation_matrix = quantum_data.drop(columns=['conc']).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=False, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.xlabel("Features")
plt.ylabel("Features")
plt.show()
```

```python

# Install TensorFlow package if not already installed
# pip install tensorflow matplotlib pandas

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np # Import numpy

# Set random seeds for reproducibility
SEED = 123
np.random.seed(SEED)
tf.random.set_seed(SEED)


# Re-initialize y_train and y_test with the original data
y_train = train_data['conc'].values
y_test = test_data['conc'].values

## Define the ANN architecture
model = Sequential([
    Dense(64, activation="relu", input_shape=(x_train.shape[1],)),  # Input layer
    Dense(32, activation="relu"),  # Hidden layer
    Dense(1)  # Output layer for regression
])

# Display the model summary
model.summary()

# Compile the model
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",  # Mean Squared Error for regression
    metrics=["mae"]  # Mean Absolute Error for evaluation
)


# Train the model
history = model.fit(
    x_train, np.asarray(y_train), # Convert y_train to a NumPy array
    epochs=50,  # Number of epochs (can be adjusted)
    batch_size=32,  # Batch size (can be adjusted)
    validation_split=0.2,  # Use 20% of training data for validation
    verbose=1  # Display progress during training
)

# Plot training history
plt.plot(history.history['loss'], label='Training Loss (MSE)') # Add back this line
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.plot(history.history['mae'], label='Training MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.title('Training History')
plt.xlabel('Epochs')
plt.ylabel('Loss/MAE')
plt.legend()
plt.show()

# Evaluate the model on the test data
results = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Loss (MSE): {results[0]}")
print(f"Test MAE: {results[1]}")

# Make predictions on test data
predictions = model.predict(x_test)

# Visualize predicted vs actual values
test_df = pd.DataFrame({
    "Actual": y_test.flatten(),
    "Predicted": predictions.flatten()
})

plt.figure(figsize=(8, 8))
plt.scatter(test_df["Actual"], test_df["Predicted"], color="blue", alpha=0.6)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color="red", linestyle="dashed")
plt.title("Actual vs Predicted Values")
plt.xlabel("Actual Concurrence")
plt.ylabel("Predicted Concurrence")
plt.grid(True)
plt.show()
```

```python
# Install TensorFlow package if not already installed
# pip install tensorflow matplotlib pandas

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import Huber # Import Huber
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np # Import numpy
from sklearn.model_selection import train_test_split # Import train_test_split


# Assuming 'quantum_data' is your DataFrame
# If not, replace it with your actual data

# Split the data (80% train, 20% test)
train_data, test_data = train_test_split(quantum_data, test_size=0.2, random_state=123) # Define train_data and test_data

# Separate features and target
x_train = train_data.drop(columns=['conc']).values
y_train = train_data['conc'].values

x_test = test_data.drop(columns=['conc']).values
y_test = test_data['conc'].values

## Define the ANN architecture
model = Sequential([
    Dense(128, activation="relu", input_shape=(x_train.shape[1],)),  # Input layer

    Dense(64, activation="relu"),  # Hidden layer

    Dense(32, activation="relu"),  # Hidden layer

    Dense(1)  # Output layer for regression
])

# Display the model summary
model.summary()


# Compile the model with Huber loss
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss=Huber(delta=1.0),
    metrics=["mae"]
)

# Train the model
history = model.fit(
    x_train, np.asarray(y_train), # Convert y_train to a NumPy array
    epochs=50,  # Number of epochs (can be adjusted)
    batch_size=32,  # Batch size (can be adjusted)
    validation_split=0.2,  # Use 20% of training data for validation
    verbose=1  # Display progress during training
)

# Plot training history
plt.plot(history.history['loss'], label='Training Loss (MSE)') # Add back this line
plt.plot(history.history['val_loss'], label='Validation Loss (MSE)')
plt.plot(history.history['mae'], label='Training MAE')
plt.plot(history.history['val_mae'], label='Validation MAE')
plt.title('Training History')
plt.xlabel('Epochs')
plt.ylabel('Loss/MAE')
plt.legend()
plt.show()

# Evaluate the model on the test data
results = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Loss (MSE): {results[0]}")
print(f"Test MAE: {results[1]}")

# Make predictions on test data
predictions = model.predict(x_test)

# Visualize predicted vs actual values
test_df = pd.DataFrame({
    "Actual": y_test.flatten(),
    "Predicted": predictions.flatten()
})

plt.figure(figsize=(8, 8))
plt.scatter(test_df["Actual"], test_df["Predicted"], color="blue", alpha=0.6)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color="red", linestyle="dashed")
plt.title("Actual vs Predicted Values")
plt.xlabel("Actual Concurrence")
plt.ylabel("Predicted Concurrence")
plt.grid(True)
plt.show()
```
